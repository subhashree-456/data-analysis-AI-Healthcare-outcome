```python
# Mean Imputation
mean_columns = [
    'Age',
    'BMI',
    'Exercise_Hours_Per_Week',
    'Health_Score_Before_AI',
    'Satisfaction_Score'
]

for col in mean_columns:
    df[col] = df[col].fillna(df[col].mean())


# Median Imputation
median_columns = [
    'AI_Health_App_Usage_Hours_Per_Week',
    'Health_Score_After_AI'
]

for col in median_columns:
    df[col] = df[col].fillna(df[col].median())
```

<h5>
Identifies all numerical columns in the dataset to perform outlier detection and analyze the distribution of continuous variables.
</h5>

```python
import math
n = len(numerical_columns)
cols = 3
rows = math.ceil(n / cols)

plt.figure(figsize=(18, rows * 4))

for i, col in enumerate(numerical_columns, 1):
    plt.subplot(rows, cols, i)
    sns.boxplot(x=df[col])
    plt.title(col, fontsize=10)
    plt.xlabel("")

plt.tight_layout()
plt.show()
```

<h5>
Creates boxplots for all numerical columns in a single figure using subplots, making it easier to identify and compare potential outliers across multiple features simultaneously.
</h5>

```python
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

<h5>

```python
Q1 = df['Improvement_Per_AI_Hour'].quantile(0.25)
Q3 = df['Improvement_Per_AI_Hour'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['Improvement_Per_AI_Hour'] = df['Improvement_Per_AI_Hour'].clip(lower, upper)
```

<h4>Question 15: Which participant profiles should healthcare companies target for maximum AI impact?</h4>

<h5>
<b>Why Horizontal Bar Chart?</b>
    
A Horizontal Bar Chart is an ideal visualization for comparing and ranking participant profiles based on their average health improvement. It clearly highlights which participant profiles benefit the most from AI-driven healthcare solutions, enabling healthcare companies to identify and prioritize the target audience for maximum AI impact.
</h5>

```python
# Create Participant Profile

df["Participant_Profile"] = (
    df["Age_Group"] + " | " +
    df["Exercise_Level"]
)

# Calculate Average Health Improvement for each Participant Profile

profile_target = (
    df.groupby("Participant_Profile")["Improvement"]
      .mean()
      .sort_values(ascending=True)
)

print(profile_target)
```

<h5>
<b>Analysis Explanation:</b>
<ul>
<li>A participant profile is created by combining the <b>Age Group</b> and <b>Exercise Level</b>.</li>
<li>The average health improvement is calculated for each participant profile.</li>
<li>The profiles are ranked based on their average improvement to identify which groups benefit the most from AI-driven healthcare solutions.</li>
</ul>
</h5>

```python
import numpy as np
import matplotlib.pyplot as plt

# Gradient Colors
colors = plt.cm.viridis(np.linspace(0.1, 0.9, len(profile_target)))

plt.figure(figsize=(11,6))

bars = plt.barh(
    profile_target.index,
    profile_target.values,
    color=colors,
    edgecolor="white",
    height=0.8
)

# Title
plt.title(
    "Participant Profiles with Highest AI Health Impact",
    fontsize=22,
    fontweight="bold",
    pad=15
)

# Axis Labels
plt.xlabel(
    "Average Health Improvement",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Participant Profile",
    fontsize=13,
    fontweight="bold"
)

# Grid
plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.4
)

# Remove top & right border
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Value Labels
for bar in bars:
    plt.text(
        bar.get_width()+0.15,
        bar.get_y()+bar.get_height()/2,
        f"{bar.get_width():.2f}",
        va="center",
        fontsize=11,
        fontweight="bold",
        color="#333333"
    )

plt.tight_layout()
plt.show()
```

<h5>
<b>Code Explanation:</b>
<ul>
<li>A new <b>Participant Profile</b> is created by combining the <b>Age Group</b> and <b>Exercise Level</b> of each participant.</li>
<li>The <b>groupby()</b> function calculates the average health improvement for every participant profile.</li>
<li>The values are sorted and displayed using a <b>Horizontal Bar Chart</b>, making it easy to compare and rank participant profiles based on their average improvement.</li>
</ul>
</h5>

<h5>
<b>Graph Explanation:</b>
<ul>
<li>Each horizontal bar represents the average health improvement achieved by a specific participant profile.</li>
<li>Profiles with longer bars demonstrate higher average improvement, indicating greater effectiveness of AI-driven healthcare interventions.</li>
<li>Healthcare companies can use these insights to prioritize participant profiles that are most likely to achieve maximum health benefits from AI applications.</li>
</ul>
</h5>

<h5>

<ul>

<li>A clustered bar chart is created to display both AI usage and health improvement side by side for each participant group.</li>

</ul>
</h5>

<h5>
